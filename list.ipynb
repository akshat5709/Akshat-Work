list method i
-append() - to add an item at the end of a list
- insert( - to add an item at any specific position)
- pop() - to delete an item using the index position
- remove() - to delete an item using the element name
- copy() - to make a copy of the list
- sort() - to arrange the items in a particular order
-extend() - to combine to list
- clear() - to delete all the elements of a list but not the list

```python
apps.remove('bgmi')
print(apps)
```

```python
apps2 = apps.copy()
print(apps2)
```

```python
apps.clear()
print(apps)
```

```python
apps2.sort()
print(apps2)
```

```python
a = ['chalo','tictok','tinder','pubg']
apps2.extend(a)
print(apps2)
```
